<h1 style="color:DodgerBlue">Generic-классы в C#</h1>

Обобщённый класс использует параметр типа `T`. Ограничение `where T : Animal` разрешает работать со свойствами и методами `Animal` и его наследников.
`AnimalCollection<Dog>` хранит собак, а `AnimalCollection<Animal>` — животных разных видов.

### Пример из урока
Для совместного хранения автомобилей и мотоциклов нужен `VehicleCollection<Vehicle>`. В `VehicleCollection<Car>` добавить `Motorcycle` нельзя: ограничение не отменяет тип `T`.

```
using System;
using System.Collections.Generic;

public class Vehicle
{
    public virtual void Drive(int speed) => Console.WriteLine($"Транспорт движется со скоростью {speed} км/ч.");
}
public class Car : Vehicle
{
    public override void Drive(int speed)
    {
        base.Drive(speed);
        Console.WriteLine($"Автомобиль едет по дороге: {speed} км/ч.");
    }
}
public class Motorcycle : Vehicle
{
    public override void Drive(int speed) => Console.WriteLine($"Мотоцикл едет: {speed} км/ч.");
}
public class VehicleCollection<T> where T : Vehicle
{
    private readonly List<T> items = new List<T>();
    public int Count => items.Count;
    public void Add(T item)
    {
        if (item == null) throw new ArgumentNullException(nameof(item));
        items.Add(item);
    }
    public bool Remove(T item) => items.Remove(item);
    public void DisplayAll()
    {
        foreach (T item in items) { item.Drive(60); }
    }
}
```

```
var vehicles = new VehicleCollection<Vehicle>();
vehicles.Add(new Car());
var motorcycle = new Motorcycle();
vehicles.Add(motorcycle);
vehicles.DisplayAll();
Console.WriteLine($"Мотоцикл удалён: {vehicles.Remove(motorcycle)}; осталось: {vehicles.Count}");
var cars = new VehicleCollection<Car>();
cars.Add(new Car());
cars.DisplayAll();
// cars.Add(new Motorcycle()); // Ошибка компиляции: ожидается Car.
```

### Задание
Создать `Animal` и 3–4 производных класса со свойствами, геттерами и сеттерами; применить generic-классы.

### Решение
`AnimalCollection<T>` поддерживает добавление, удаление и вывод элементов. Полиморфные вызовы выполняются без приведения типов.

```
public class Animal
{
    public string Name { get; set; }
    private int age;
    public int Age
    {
        get => age;
        set => age = value >= 0 ? value : throw new ArgumentOutOfRangeException(nameof(value));
    }
    public Animal(string name, int age) { Name = name; Age = age; }
    public virtual void MakeSound(int volume) => Console.WriteLine($"{Name}: звук животного, громкость {volume}.");
    public virtual void DisplayInfo() => Console.WriteLine($"{Name}, возраст: {Age}");
}
public class Dog : Animal
{
    public string Breed { get; set; }
    public Dog(string name, int age, string breed) : base(name, age) => Breed = breed;
    public override void MakeSound(int volume) => Console.WriteLine($"{Name}: Гав! Громкость {volume}.");
    public override void DisplayInfo() { base.DisplayInfo(); Console.WriteLine($"Порода: {Breed}"); }
}
public class Cat : Animal
{
    public string Color { get; set; }
    public Cat(string name, int age, string color) : base(name, age) => Color = color;
    public override void MakeSound(int volume) => Console.WriteLine($"{Name}: Мяу! Громкость {volume}.");
    public override void DisplayInfo() { base.DisplayInfo(); Console.WriteLine($"Окрас: {Color}"); }
}
public class Bird : Animal
{
    public double WingSpan { get; set; }
    public Bird(string name, int age, double wingSpan) : base(name, age) => WingSpan = wingSpan;
    public override void MakeSound(int volume) => Console.WriteLine($"{Name}: Чирик! Громкость {volume}.");
    public override void DisplayInfo() { base.DisplayInfo(); Console.WriteLine($"Размах крыльев: {WingSpan} см"); }
}
public class AnimalCollection<T> where T : Animal
{
    private readonly List<T> items = new List<T>();
    public int Count => items.Count;
    public void Add(T item)
    {
        if (item == null) throw new ArgumentNullException(nameof(item));
        items.Add(item);
    }
    public bool Remove(T item) => items.Remove(item);
    public void DisplayAll()
    {
        foreach (T item in items) { item.DisplayInfo(); item.MakeSound(5); }
    }
}
```

```
var animals = new AnimalCollection<Animal>();
var dog = new Dog("Рекс", 3, "Овчарка");
var cat = new Cat("Мурка", 2, "Рыжий");
var bird = new Bird("Кеша", 1, 25);
dog.Age = 4;
animals.Add(dog);
animals.Add(cat);
animals.Add(bird);
animals.DisplayAll();
Console.WriteLine($"Кошка удалена: {animals.Remove(cat)}; осталось: {animals.Count}");
animals.DisplayAll();
var dogs = new AnimalCollection<Dog>();
dogs.Add(dog);
dogs.DisplayAll();
// dogs.Add(cat); // Ошибка компиляции: коллекция принимает только Dog.
```
