<h1 style="color:DodgerBlue">Перекрытие и перегрузка методов</h1>

В уроке «перекрытие» используется в значении переопределения (`override`). Сигнатура сохраняется, поведение меняется в наследнике. `base` вызывает реализацию родителя.
Перегрузка — методы с одним именем и разными параметрами. Компилятор выбирает наиболее подходящую перегрузку; неоднозначный вызов приводит к ошибке компиляции.
Скрытие через `new` — отдельный механизм, здесь используется переопределение.

### Пример из урока

```
using System;
using System.Collections.Generic;

public class Vehicle
{
    public void Drive(string mode) => Console.WriteLine($"Режим движения: {mode}");
    public void Drive(bool isManual) => Console.WriteLine(isManual ? "Ручное переключение передач" : "Автоматическое переключение передач");
    public virtual void Drive(int speed) => Console.WriteLine($"Транспорт движется со скоростью {speed} км/ч.");
}
public class Car : Vehicle
{
    public override void Drive(int speed)
    {
        base.Drive(speed);
        Console.WriteLine($"Автомобиль едет по дороге: {speed} км/ч.");
    }
}
public class Motorcycle : Vehicle
{
    public override void Drive(int speed) => Console.WriteLine($"Мотоцикл едет: {speed} км/ч.");
}
```

```
Vehicle[] vehicles = { new Vehicle(), new Car(), new Motorcycle() };
foreach (Vehicle vehicle in vehicles) vehicle.Drive(60);
vehicles[1].Drive("Эко");
vehicles[1].Drive(true);
vehicles[2].Drive(false);
```

### Задание
Создать `Animal` и 3–4 производных класса со свойствами, геттерами и сеттерами. Использовать перекрытие и перегрузку методов.

### Решение
Три наследника переопределяют `MakeSound(int)` и расширяют `DisplayInfo()` через `base`.
Перегрузки `MakeSound()`, `MakeSound(int)` и `MakeSound(string)` вызываются ниже. Даже вызов без параметров обращается к переопределению `MakeSound(int)` реального типа.

```
public class Animal
{
    public string Name { get; set; }
    private int age;
    public int Age
    {
        get => age;
        set => age = value >= 0 ? value : throw new ArgumentOutOfRangeException(nameof(value));
    }
    public Animal(string name, int age) { Name = name; Age = age; }
    // Перегрузки различаются числом и типами параметров.
    public void MakeSound() => MakeSound(5);
    public void MakeSound(string mood)
    {
        Console.WriteLine($"Настроение: {mood}");
        MakeSound(3);
    }
    public virtual void MakeSound(int volume) => Console.WriteLine($"{Name}: звук животного, громкость {volume}.");
    public virtual void DisplayInfo() => Console.WriteLine($"{Name}, возраст: {Age}");
}
public class Dog : Animal
{
    public string Breed { get; set; }
    public Dog(string name, int age, string breed) : base(name, age) => Breed = breed;
    public override void MakeSound(int volume) => Console.WriteLine($"{Name}: Гав! Громкость {volume}.");
    public override void DisplayInfo() { base.DisplayInfo(); Console.WriteLine($"Порода: {Breed}"); }
}
public class Cat : Animal
{
    public string Color { get; set; }
    public Cat(string name, int age, string color) : base(name, age) => Color = color;
    public override void MakeSound(int volume) => Console.WriteLine($"{Name}: Мяу! Громкость {volume}.");
    public override void DisplayInfo() { base.DisplayInfo(); Console.WriteLine($"Окрас: {Color}"); }
}
public class Bird : Animal
{
    public double WingSpan { get; set; }
    public Bird(string name, int age, double wingSpan) : base(name, age) => WingSpan = wingSpan;
    public override void MakeSound(int volume) => Console.WriteLine($"{Name}: Чирик! Громкость {volume}.");
    public override void DisplayInfo() { base.DisplayInfo(); Console.WriteLine($"Размах крыльев: {WingSpan} см"); }
}
```

```
Animal[] animals = { new Dog("Рекс", 3, "Овчарка"), new Cat("Мурка", 2, "Рыжий"), new Bird("Кеша", 1, 25) };
animals[0].Age = 4;
foreach (Animal animal in animals)
{
    animal.DisplayInfo();
    animal.MakeSound(5);
}
Console.WriteLine("Перегрузки:");
foreach (Animal animal in animals)
{
    animal.MakeSound();
    animal.MakeSound(8);
    animal.MakeSound("Радостное");
}
```
