### Вариант задания №18

**Студент:** Чирченко Максим

Создать базовый класс `Review` для представления отзывов о продуктах и услугах. Класс содержит идентификатор отзыва, текст и рейтинг, а также методы `DisplayReview()`, `RateProduct()` и `GetReviewDetails()`. Производные классы демонстрируют наследование и полиморфизм: `ServiceReview` добавляет дату посещения, `ProductReview` — идентификатор продукта, а `ServiceTimeReview` — время начала услуги.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;

public class Review
{
    public int ReviewId { get; }
    public string Text { get; }
    public int Rating { get; private set; }
    public Review(int reviewId, string text, int rating) { ReviewId = reviewId; Text = text; Rating = Math.Clamp(rating, 1, 5); }
    public virtual void DisplayReview() => Console.WriteLine($"Отзыв #{ReviewId}: {Text} ({Rating}/5)");
    public virtual void RateProduct(int rating) { Rating = Math.Clamp(rating, 1, 5); }
    public virtual string GetReviewDetails() => $"#{ReviewId}, рейтинг {Rating}/5: {Text}";
}
public class ServiceReview : Review
{
    public DateTime VisitDate { get; }
    public ServiceReview(int id, string text, int rating, DateTime visitDate) : base(id, text, rating) => VisitDate = visitDate;
    public override void DisplayReview() => Console.WriteLine($"Сервисный отзыв #{ReviewId} от {VisitDate:dd.MM.yyyy}: {Text} ({Rating}/5)");
    public override string GetReviewDetails() => $"{base.GetReviewDetails()}, дата посещения: {VisitDate:dd.MM.yyyy}";
}
public class ProductReview : Review
{
    public string ProductId { get; }
    public ProductReview(int id, string text, int rating, string productId) : base(id, text, rating) => ProductId = productId;
    public override void RateProduct(int rating) { base.RateProduct(rating); Console.WriteLine($"Товар {ProductId} получил рейтинг {Rating}/5"); }
    public override string GetReviewDetails() => $"{base.GetReviewDetails()}, товар: {ProductId}";
}
public class ServiceTimeReview : ServiceReview
{
    public TimeSpan StartTime { get; }
    public ServiceTimeReview(int id, string text, int rating, DateTime date, TimeSpan startTime) : base(id, text, rating, date) => StartTime = startTime;
    public override string GetReviewDetails() => $"{base.GetReviewDetails()}, начало услуги: {StartTime:hh\\:mm}";
}
public class Program { public static void Main() { Review[] reviews = { new ServiceReview(1, "Вежливый персонал", 5, new DateTime(2026, 9, 18)), new ProductReview(2, "Качество соответствует описанию", 4, "P-018"), new ServiceTimeReview(3, "Услуга началась вовремя", 5, new DateTime(2026, 9, 19), new TimeSpan(14, 30, 0)) };
foreach (Review review in reviews) { review.DisplayReview(); Console.WriteLine(review.GetReviewDetails()); }
((ProductReview)reviews[1]).RateProduct(5); } }

```
