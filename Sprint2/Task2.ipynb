<h1 style="color:DodgerBlue">Множественное наследование. Абстрактные классы и интерфейсы</h1>

C# допускает один базовый класс и реализацию нескольких интерфейсов. Так объединяют несколько контрактов поведения.
Абстрактный класс хранит общее состояние и реализацию; его экземпляр создать нельзя. Абстрактный метод реализуется в производном классе.
Интерфейс задаёт контракт свойств и методов. В примере `Vehicle` реализует `IVehicle`, а его наследники дополнительно реализуют `IMotorized`.

### Пример из урока: Vehicle, Car, Motorcycle и интерфейсы

```
using System;

public interface IVehicle
{
    string Make { get; set; }
    string Model { get; set; }
    void DisplayInfo();
}
public interface IMotorized
{
    string EngineType { get; set; }
    void StartEngine();
}
public abstract class Vehicle : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    public int Speed { get; set; }
    protected Vehicle(string make, string model) { Make = make; Model = model; }
    public virtual void Move() => Console.WriteLine($"{Make} {Model}: скорость {Speed} км/ч");
    public abstract void DisplayInfo();
}
public class Car : Vehicle, IMotorized
{
    public int Doors { get; set; }
    public string EngineType { get; set; }
    public Car(string make, string model, int doors, string engineType) : base(make, model)
    { Doors = doors; EngineType = engineType; }
    public void StartEngine() => Console.WriteLine($"Двигатель автомобиля запущен: {EngineType}");
    public override void DisplayInfo() => Console.WriteLine($"Автомобиль {Make} {Model}, дверей: {Doors}");
}
public class Motorcycle : Vehicle, IMotorized
{
    public string EngineType { get; set; }
    public Motorcycle(string make, string model, string engineType) : base(make, model) => EngineType = engineType;
    public void StartEngine() => Console.WriteLine($"Двигатель мотоцикла запущен: {EngineType}");
    public override void DisplayInfo() => Console.WriteLine($"Мотоцикл {Make} {Model}, двигатель: {EngineType}");
    public override void Move() { base.Move(); Console.WriteLine("Мотоцикл движется на двух колёсах"); }
}
Vehicle[] vehicles = { new Car("Toyota", "Corolla", 4, "Бензиновый") { Speed = 60 }, new Motorcycle("Honda", "CBR600RR", "Inline Four") { Speed = 80 } };
foreach (Vehicle vehicle in vehicles)
{
    IVehicle info = vehicle;
    info.DisplayInfo();
    ((IMotorized)vehicle).StartEngine();
    vehicle.Move();
}
```

### Задание
Создать базовый класс `Animal` и 3–4 производных класса со свойствами, геттерами и сеттерами; реализовать множественное наследование через интерфейсы.

### Решение
`Animal` — абстрактный класс. `Dog : Animal, IPet, ITrainable`, `Cat : Animal, IPet`, `Bird : Animal, IPet, IFlyable`.
Вызовы через `IPet`, `ITrainable`, `IFlyable` показывают работу контрактов, а массив `Animal[]` — полиморфизм.

```
public interface IPet
{
    string Owner { get; set; }
    void Play();
}
public interface ITrainable
{
    void ExecuteCommand(string command);
}
public interface IFlyable
{
    double WingSpan { get; set; }
    void Fly();
}
public abstract class Animal
{
    public string Name { get; set; }
    private int age;
    public int Age
    {
        get => age;
        set => age = value >= 0 ? value : throw new ArgumentOutOfRangeException(nameof(value));
    }
    protected Animal(string name, int age) { Name = name; Age = age; }
    public virtual void DisplayInfo() => Console.WriteLine($"{Name}, возраст: {Age}");
    public abstract void MakeSound();
}
public class Dog : Animal, IPet, ITrainable
{
    public string Owner { get; set; }
    public string Breed { get; set; }
    public Dog(string name, int age, string owner, string breed) : base(name, age) { Owner = owner; Breed = breed; }
    public override void MakeSound() => Console.WriteLine("Гав!");
    public void Play() => Console.WriteLine($"{Name} играет с хозяином {Owner}");
    public void ExecuteCommand(string command) => Console.WriteLine($"{Name} выполняет команду: {command}");
}
public class Cat : Animal, IPet
{
    public string Owner { get; set; }
    public string Color { get; set; }
    public Cat(string name, int age, string owner, string color) : base(name, age) { Owner = owner; Color = color; }
    public override void MakeSound() => Console.WriteLine("Мяу!");
    public void Play() => Console.WriteLine($"{Name} играет с клубком у хозяина {Owner}");
}
public class Bird : Animal, IPet, IFlyable
{
    public string Owner { get; set; }
    public double WingSpan { get; set; }
    public Bird(string name, int age, string owner, double wingSpan) : base(name, age) { Owner = owner; WingSpan = wingSpan; }
    public override void MakeSound() => Console.WriteLine("Чирик!");
    public void Play() => Console.WriteLine($"{Name} играет с хозяином {Owner}");
    public void Fly() => Console.WriteLine($"{Name} летит, размах крыльев {WingSpan} см");
}
Animal[] animals = { new Dog("Рекс", 3, "Максим", "Овчарка"), new Cat("Мурка", 2, "Анна", "Рыжий"), new Bird("Кеша", 1, "Олег", 25) };
animals[0].Age = 4;
foreach (Animal animal in animals)
{
    animal.DisplayInfo();
    animal.MakeSound();
    if (animal is IPet pet) pet.Play();
    if (animal is ITrainable trainable) trainable.ExecuteCommand("Сидеть");
    if (animal is IFlyable flyable) flyable.Fly();
}
```

### Различия

| Возможность | Абстрактный класс | Интерфейс |
|---|---|---|
| Поля экземпляра и общее состояние | Да | Нет |
| Конструктор экземпляра | Да, вызывается наследниками | Нет |
| Количество у класса | Один базовый класс | Несколько интерфейсов |
| Назначение | Общая основа родственных объектов | Контракт поведения |

В этих примерах интерфейсы содержат только контракты; реализация находится в классах.
