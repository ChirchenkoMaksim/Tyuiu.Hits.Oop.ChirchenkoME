<h1 style="color:DodgerBlue">Сложное (многоуровневое) наследование</h1>

При многоуровневом наследовании производный класс становится базовым для следующего класса.
Общие свойства передаются по всей цепочке; конструкторы вызываются от базового к производному.

Пример: `Vehicle → Car → SportsCar`; отдельная ветвь: `Vehicle → Motorcycle`.

### Пример из урока: транспорт и спортивный автомобиль

```
using System;

public class Vehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    public Vehicle(string make, string model) { Make = make; Model = model; }
    public virtual void DisplayInfo() => Console.WriteLine($"Транспорт: {Make} {Model}");
}
public class Car : Vehicle
{
    public int Doors { get; set; }
    public Car(string make, string model, int doors) : base(make, model) => Doors = doors;
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Количество дверей: {Doors}");
    }
}
public class Motorcycle : Vehicle
{
    public string EngineType { get; set; }
    public Motorcycle(string make, string model, string engineType) : base(make, model) => EngineType = engineType;
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Тип двигателя: {EngineType}");
    }
}
public class SportsCar : Car
{
    public bool IsConvertible { get; set; }
    public SportsCar(string make, string model, int doors, bool isConvertible) : base(make, model, doors) => IsConvertible = isConvertible;
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Откидной верх: {IsConvertible}");
    }
}
Vehicle[] vehicles = { new Car("Toyota", "Corolla", 4), new SportsCar("Mazda", "MX-5", 2, true), new Motorcycle("Honda", "CBR600RR", "Inline Four") };
foreach (Vehicle vehicle in vehicles) vehicle.DisplayInfo();
```

### Задание
Создать `Animal` и 3–4 производных класса со свойствами, геттерами и сеттерами; реализовать сложное наследование.

### Решение
Четыре производных класса: `Mammal`, `Dog`, `Cat`, `Bird`.
Цепочки: `Animal → Mammal → Dog`, `Animal → Mammal → Cat`; `Bird` наследует `Animal`.
Собаки и кошки наследуют также свойство `FurColor` и метод `FeedMilk` промежуточного класса.

```
public class Animal
{
    public string Name { get; set; }
    private int age;
    public int Age
    {
        get => age;
        set => age = value >= 0 ? value : throw new ArgumentOutOfRangeException(nameof(value));
    }
    public Animal(string name, int age) { Name = name; Age = age; }
    public virtual void DisplayInfo() => Console.WriteLine($"{Name}, возраст: {Age}");
    public virtual void MakeSound() => Console.WriteLine("Животное издаёт звук");
}
public class Mammal : Animal
{
    public string FurColor { get; set; }
    public Mammal(string name, int age, string furColor) : base(name, age) => FurColor = furColor;
    public void FeedMilk() => Console.WriteLine($"{Name} кормит детёнышей молоком");
    public override void DisplayInfo() { base.DisplayInfo(); Console.WriteLine($"Цвет шерсти: {FurColor}"); }
}
public class Dog : Mammal
{
    public string Breed { get; set; }
    public Dog(string name, int age, string breed) : base(name, age, "Чёрный") => Breed = breed;
    public override void DisplayInfo() { base.DisplayInfo(); Console.WriteLine($"Порода: {Breed}"); }
    public override void MakeSound() => Console.WriteLine("Гав!");
}
public class Cat : Mammal
{
    public string Color { get; set; }
    public Cat(string name, int age, string color) : base(name, age, color) => Color = color;
    public override void DisplayInfo() { base.DisplayInfo(); Console.WriteLine($"Окрас: {Color}"); }
    public override void MakeSound() => Console.WriteLine("Мяу!");
}
public class Bird : Animal
{
    public double WingSpan { get; set; }
    public Bird(string name, int age, double wingSpan) : base(name, age) => WingSpan = wingSpan;
    public override void DisplayInfo() { base.DisplayInfo(); Console.WriteLine($"Размах крыльев: {WingSpan} см"); }
    public override void MakeSound() => Console.WriteLine("Чирик!");
}

Animal[] animals = { new Dog("Рекс", 3, "Овчарка"), new Cat("Мурка", 2, "Рыжий"), new Bird("Кеша", 1, 25) };
animals[0].Age = 4;
foreach (Animal animal in animals) { animal.DisplayInfo(); animal.MakeSound(); }

((Dog)animals[0]).FeedMilk();
((Cat)animals[1]).FeedMilk();
```
